**وظیفه:** محاسبه میانگین، میانه، واریانس و انحراف معیار برای ستون‌های `nodes_minimax`، `nodes_alphabeta` و `moves`.

**داده:** `ttt_games.csv` (۲۰۰ بازی، ۱۰ ستون).

## ۱. بارگذاری داده

```python
import pandas as pd

g = pd.read_csv("ttt_games.csv", dtype={"start_board": str})
print("شکل داده:", g.shape)
print("مقادیر گم‌شده:", int(g.isnull().sum().sum()))
```

## ۲. آمار توصیفی
واریانس و انحراف معیار با `pandas` به‌صورت پیش‌فرض **نمونه‌ای (n−1)** حساب می‌شوند. نسخه جامعه (n) هم در سلول بعد آمده است.

```python
cols = ["nodes_minimax", "nodes_alphabeta", "moves"]

stats = pd.DataFrame({
    "mean":   g[cols].mean(),
    "median": g[cols].median(),
    "var":    g[cols].var(),     # واریانس نمونه (n-1)، پیش‌فرض pandas
    "std":    g[cols].std(),     # انحراف معیار نمونه (n-1)
}).round(2)
print(stats.to_string())
```

```python
pop = pd.DataFrame({
    "var_population": g[cols].var(ddof=0),   # واریانس جامعه (n)
    "std_population": g[cols].std(ddof=0),
}).round(2)
print(pop.to_string())
```

## ۳. نسبت میانگین به میانه (بررسی کجی)

```python
print((g[cols].mean() / g[cols].median()).round(2).to_string())
```

## تفسیر کوتاه (بر اساس خروجی بالا)

- **`nodes_minimax`:** میانگین 120,069 ولی میانه فقط 6,812 است و انحراف معیار (216,379) از میانگین بزرگ‌تر است. توزیع به‌شدت راست‌کج است و چند بازی پرهزینه (شروع از تخته خالی) میانگین را بالا می‌برند؛ برای توصیف حالت معمول میانه مناسب‌تر است.
- **`nodes_alphabeta`:** همین الگو در مقیاس کوچک‌تر؛ میانگین 4,382 و میانه 427.
- **`moves`:** میانگین (4.56) و میانه (4.5) تقریباً برابرند و انحراف معیار 1.64 است؛ توزیع نسبتاً متقارن و کم‌پراکندگی است.
- **نکته:** ۴۰ بازی از ۲۰۰ بازی از تخته خالی شروع شده‌اند و ردیف‌های تکراری وجود دارد؛ این آمار روی همه ۲۰۰ ردیف و بدون حذف تکراری‌ها حساب شده است.

## ۴. بررسی QA: کاهش کل گره‌ها

```python
total_mm = g.nodes_minimax.sum()
total_ab = g.nodes_alphabeta.sum()
print(f"مجموع گره‌های Minimax:    {total_mm:,}")
print(f"مجموع گره‌های Alpha-Beta: {total_ab:,}")
print(f"کاهش کل گره‌ها: {(1 - total_ab / total_mm) * 100:.2f}%")
```
